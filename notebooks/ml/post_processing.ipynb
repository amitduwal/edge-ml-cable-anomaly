```python
import h5py
import numpy as np
from scipy.stats import f_oneway
```

```python
# Step 1: Load the data
mat_file = r'E:\Thesis\thesis_code\data\rp\100Mbps_features.h5'   # change path
```

```python
with h5py.File(mat_file, 'r') as f:
    X = np.array(f['features'])          # Features
    y = np.array(f['label_type'])        # Labels
    z = np.array(f['label_length'])     # Length labels (if needed)

print("Feature shape:", X.shape)
print("Label shape:", y.shape)
```

```python
n_features = X.shape[1]

f_values = []
p_values = []

# Run ANOVA feature-wise
for i in range(n_features):
    groups = []

    # split feature i by class
    for cls in np.unique(y):
        groups.append(X[y == cls, i])

    # one-way ANOVA
    f_val, p_val = f_oneway(*groups)

    f_values.append(f_val)
    p_values.append(p_val)

f_values = np.array(f_values)
p_values = np.array(p_values)

# Rank features by importance (higher F = more discriminative)
ranked_idx = np.argsort(f_values)[::-1]

print("Top 10 features by ANOVA F-score:")
print(ranked_idx[:30])
print(f_values[ranked_idx[:30]])
```
