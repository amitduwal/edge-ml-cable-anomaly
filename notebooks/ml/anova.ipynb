```python
import h5py
import numpy as np
```

```python
mat_file = r'E:\Thesis\thesis_code\data\rp\100Mbps_features.h5'   # change path
```

```python
with h5py.File(mat_file, 'r') as f:
    X = np.array(f['features'])          # Features
    y = np.array(f['label_type'])        # Labels
    z = np.array(f['label_length'])     # Length labels (if needed)

print("Feature shape:", X.shape)
print("Label shape:", y.shape)
```

```python
import pandas as pd
from sklearn.feature_selection import f_classif
```

```python

# ANOVA feature ranking
f_scores, p_values = f_classif(X, y)
```

```python
# Create ranking table
feature_ranking = pd.DataFrame({
    "Feature": [f"F{i+1}" for i in range(X.shape[1])],
    "F_score": f_scores,
    "p_value": p_values
})

feature_ranking = feature_ranking.sort_values(
    by="F_score",
    ascending=False
)

print(feature_ranking)
```

```python
feature_names = [
    # Time-domain features
    "std",
    "skewness",
    "rms",
    "mean",
    "kurtosis",
    "peak",
    "shape_factor",
    "impulse_factor",
    "crest_factor",
    "clearance_factor",

    # Frequency-domain features
    "freq_mag_0",
    "freq_val_0",
    "freq_mag_1",
    "freq_val_1",
    "freq_mag_2",
    "freq_val_2",
    "freq_mag_3",
    "freq_val_3",
    "freq_mag_4",
    "freq_val_4",
    "freq_mag_5",
    "freq_val_5",
    "freq_mag_6",
    "freq_val_6",
    "freq_mag_7",
    "freq_val_7",
    "freq_mag_8",
    "freq_val_8",
    "freq_mag_9",
    "freq_val_9",
]

f_scores, p_values = f_classif(X, y)

feature_ranking = pd.DataFrame({
    "Feature": feature_names,
    "F_score": f_scores,
    "p_value": p_values
})

feature_ranking = feature_ranking.sort_values(
    by="F_score",
    ascending=False
)

print("\n=== ANOVA Feature Ranking ===")
print(feature_ranking.to_string(index=False))

# Top 10 features only
print("\n=== Top 10 Features ===")
print(feature_ranking.head(10).to_string(index=False))
```

```python
# 1. Define time-domain feature names in order of extraction
time_features = [
    'std', 
    'skewness', 
    'rms', 
    'mean', 
    'kurtosis', 
    'peak',
    'shape_factor', 
    'impulse_factor', 
    'crest_factor', 
    'clearance_factor'
]

# 2. Define frequency-domain feature names (interleaved magnitude and value for top 10 harmonics)
freq_features = []
for i in range(10):
    freq_features.append(f'freq_mag_{i}')
    freq_features.append(f'freq_val_{i}')

# 3. Combine into a single ordered list
feature_names = time_features + freq_features

# 4. Create the 0-29 index map dictionary
feature_map = {idx: name for idx, name in enumerate(feature_names)}

# Verification: Print the map
print("Index Map of Features:")
for idx, name in feature_map.items():
    print(f"{idx:2d} -> {name}")
```

```python
from enum import Enum
import h5py
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns


class ChannelCondition(Enum):
    NORMAL = 0
    AIR = 1
    WATER = 2
    TAPPED = 3


label_map = {
    ChannelCondition.NORMAL.value: "Normal",
    ChannelCondition.AIR.value: "Air",
    ChannelCondition.WATER.value: "Water",
    ChannelCondition.TAPPED.value: "Tapped",
}

# --- 1. Load and process 10BASE-T Data ---
with h5py.File(r"E:\Thesis\thesis_code\data\rp\10Mbps_features.h5", "r") as f:
    X_10bt = np.array(f["features"])  # Features shape: (N, features)
    y_10bt = np.array(f["label_type"]).flatten()  # Flatten to ensure 1D array

df_10bt = pd.DataFrame(
    {
        "Class": [label_map[label] for label in y_10bt],
        "Peak Value(V)": X_10bt[:, 5],  # Extract peak feature (index 5)
        "Standard": "10BASE-T",
    }
)

# --- 2. Load and process 10BASE-TX Data ---
with h5py.File(r"E:\Thesis\thesis_code\data\rp\100Mbps_features.h5", "r") as f:
    X_100tx = np.array(f["features"])
    y_100tx = np.array(f["label_type"]).flatten()

df_100tx = pd.DataFrame(
    {
        "Class": [label_map[label] for label in y_100tx],
        "Peak Value(V)": X_100tx[:, 5],
        "Standard": "100BASE-TX",
    }
)

# --- 3. Combine DataFrames ---
df_combined = pd.concat([df_10bt, df_100tx], ignore_index=True)

# --- 4. Plot Grouped Boxplot ---
plt.figure(figsize=(10, 6))

# Set all matplotlib fonts to 16
plt.rcParams.update({"font.size": 20})


sns.boxplot(
    data=df_combined,
    x="Class",
    y="Peak Value(V)",
    hue="Standard",
    palette="Set2",
    width=0.6,
)

plt.xlabel("Class", fontsize=20)
plt.ylabel("Peak Value(V)", fontsize=20)

plt.xticks(fontsize=16)
plt.yticks(fontsize=16)

plt.legend(
    title="Ethernet Standard",
    fontsize=16,
    title_fontsize=16
)

plt.grid(True, axis="y", alpha=0.3)

plt.tight_layout()
plt.show()
```
